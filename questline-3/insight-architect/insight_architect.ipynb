```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "StudentsPerformance.csv",
    encoding_errors="ignore"
)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=10)

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_math = df.groupby("gender")["math score"].mean()

plt.figure(figsize=(7, 5))

plt.bar(gender_math.index, gender_math.values)

plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```

```python
completed = df[df["test preparation course"] == "completed"]["math score"]
none = df[df["test preparation course"] == "none"]["math score"]

plt.figure(figsize=(8, 5))

plt.boxplot([completed, none], labels=["Completed", "None"])

plt.title("Math Scores by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Math Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["math score"], df["reading score"])

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.show()
```

```python
scores = df[["math score", "reading score", "writing score"]]

correlation = scores.corr()

plt.figure(figsize=(7, 5))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Between Exam Scores")

plt.show()
```
